```python
import pandas as pd
```

```python
df_fold_0 = pd.read_csv('baselines/RT_fold_0_summary .csv')
df_fold_1 = pd.read_csv('baselines/RT_fold_1_summary .csv')
df_fold_2 = pd.read_csv('baselines/RT_fold_2_summary .csv')
df_fold_3 = pd.read_csv('baselines/RT_fold_3_summary .csv')
df_fold_4 = pd.read_csv('baselines/RT_fold_4_summary .csv')
df_fold_0['fold'] = 0
df_fold_1['fold'] = 1
df_fold_2['fold'] = 2
df_fold_3['fold'] = 3
df_fold_4['fold'] = 4
df_base = pd.concat([df_fold_0, df_fold_1, df_fold_2, df_fold_3, df_fold_4], axis=0)
df_base.drop(columns=[col for col in df_base.columns if col.endswith('e_mean')], inplace=True)
df_base.drop(columns=[col for col in df_base.columns if col.endswith('_std')], inplace=True)
df_rt = df_base.groupby('model').mean().reset_index().copy()
df_rt.drop(columns=['seeds', 'fold'], inplace = True)
df_rt.columns = df_rt.columns.str.replace('_mean', '')
```

```python
df_ae = pd.read_csv('new_ghcf/rotten_tomatoes/cf_autoencoder_summary.csv')
df_ae['model'] = 'Auto Encoder'
df_ae['dataset'] = 'Rotten Tomatoes'
df_hybrid = pd.read_csv('new_ghcf/rotten_tomatoes/gated_hybrid_summary.csv')
df_hybrid['model'] = 'Hybrid'
df_hybrid['dataset'] = 'Rotten Tomatoes'
df_ghc2f = pd.read_csv('new_ghcf/rotten_tomatoes/ghc2f_summary.csv')
df_ghc2f['model'] = 'GHC2F'
df_ghc2f['dataset'] = 'Rotten Tomatoes'
```

```python
df_rt['dataset'] = 'Rotten Tomatoes'
cols = df_rt.columns
```

```python
def transpose_df(df):
    df_t = df.T
    metricas = df_t.iloc[0].values
    valores  = df_t.iloc[1].values
    df_novo = pd.DataFrame([valores], columns=metricas)

    df_novo.insert(0, 'model',   df_t.iloc[4, 0])
    df_novo.insert(1, 'dataset', df_t.iloc[5, 0])

    cols_metricas = df_novo.columns.drop(['model', 'dataset'])
    df_novo[cols_metricas] = df_novo[cols_metricas].apply(pd.to_numeric, errors='coerce')

    df_novo = df_novo.reset_index(drop=True)
    # display(df_novo)
    return df_novo
```

```python
df_novo = transpose_df(df_ae)
```

```python
df_rt = pd.concat([df_rt,transpose_df(df_ae)[cols], transpose_df(df_hybrid)[cols], transpose_df(df_ghc2f)[cols]], axis=0)
```
